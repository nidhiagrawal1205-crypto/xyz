```python
import pandas as pd
```

```python
goa = pd.read_csv('/company_master_goa.csv.csv')
chattisgarh = pd.read_csv('/company_master_chattisgarh.csv (2).csv')
madhyaPradesh = pd.read_csv('/company_master_mp.csv.csv')
rajasthan = pd.read_csv('/company_master_rajasthan.csv.csv')
```

```python
combined_df = pd.concat([goa,chattisgarh,madhyaPradesh , rajasthan],ignore_index=True)
```

```python
combined_df.shape
```

```python
combined_df['Company State'].value_counts()
```

```python
combined_df['Company State'] = combined_df['Company State'].str.strip().str.upper()
```

```python
combined_df['Company Registration Date'] = pd.to_datetime(combined_df['Company Registration Date'])
```
